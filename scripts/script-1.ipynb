### **Librerías**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("../dataset/Pokemon.xlsx")
df.head()
```

```python
df.info()
```

## Análisis descriptivos

```python
import sys
print(sys.executable)
```

```python
plt

# =========================================================
# CONFIGURACIÓN GENERAL
# =========================================================

sns.set_style("whitegrid")

plt.rcParams["figure.figsize"] = (10, 6)

# =========================================================
# INFORMACIÓN GENERAL
# =========================================================

print("=" * 60)
print("INFORMACIÓN GENERAL DEL DATASET")
print("=" * 60)

print(df.info())

print("\n")
print("=" * 60)
print("ESTADÍSTICAS DESCRIPTIVAS")
print("=" * 60)

print(df.describe())

# =========================================================
# 1. DISTRIBUCIÓN DE POKÉMON POR TIPO PRINCIPAL
# =========================================================

plt.figure(figsize=(14,6))

sns.countplot(
    data=df,
    x="Type 1",
    order=df["Type 1"].value_counts().index,
    palette="Set2"
)

plt.title("Cantidad de Pokémon por Tipo Principal", fontsize=16)
plt.xlabel("Tipo Principal")
plt.ylabel("Cantidad")

plt.xticks(rotation=45)

plt.show()

# =========================================================
# 2. DISTRIBUCIÓN DE POKÉMON LEGENDARIOS
# =========================================================

plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x="Legendary",
    palette=["#3498db", "#e74c3c"]
)

plt.title("Distribución de Pokémon Legendarios", fontsize=16)
plt.xlabel("Legendario")
plt.ylabel("Cantidad")

plt.show()

# =========================================================
# 3. DISTRIBUCIÓN DEL TOTAL DE ESTADÍSTICAS
# =========================================================

plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="Total",
    bins=30,
    kde=True,
    color="purple"
)

plt.title("Distribución del Total de Estadísticas", fontsize=16)
plt.xlabel("Total")
plt.ylabel("Frecuencia")

plt.show()

# =========================================================
# 4. BOXPLOT DEL TOTAL POR GENERACIÓN
# =========================================================

plt.figure(figsize=(12,6))

sns.boxplot(
    data=df,
    x="Generation",
    y="Total",
    palette="viridis"
)

plt.title("Distribución del Total por Generación", fontsize=16)
plt.xlabel("Generación")
plt.ylabel("Total")

plt.show()

# =========================================================
# 5. DISTRIBUCIÓN DE HP
# =========================================================

plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="HP",
    bins=30,
    kde=True,
    color="green"
)

plt.title("Distribución de HP", fontsize=16)
plt.xlabel("HP")
plt.ylabel("Frecuencia")

plt.show()

# =========================================================
# 6. ATAQUE VS DEFENSA
# =========================================================

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="Attack",
    y="Defense",
    hue="Legendary",
    palette="coolwarm",
    s=100
)

plt.title("Relación entre Ataque y Defensa", fontsize=16)
plt.xlabel("Ataque")
plt.ylabel("Defensa")

plt.show()

# =========================================================
# 7. VELOCIDAD POR TIPO PRINCIPAL
# =========================================================

plt.figure(figsize=(14,6))

sns.boxplot(
    data=df,
    x="Type 1",
    y="Speed",
    palette="tab20"
)

plt.title("Distribución de Velocidad por Tipo Principal", fontsize=16)
plt.xlabel("Tipo Principal")
plt.ylabel("Velocidad")

plt.xticks(rotation=45)

plt.show()

# =========================================================
# 8. MAPA DE CORRELACIÓN
# =========================================================

plt.figure(figsize=(10,8))

corr = df.select_dtypes(include=["int64"]).corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="RdYlBu",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Mapa de Correlación de Variables Numéricas", fontsize=16)

plt.show()

# =========================================================
# 9. ESTADÍSTICAS PROMEDIO POR TIPO
# =========================================================

stats_tipo = df.groupby("Type 1")[
    ["HP", "Attack", "Defense", "Sp. Atk", "Sp. Def", "Speed"]
].mean()

stats_tipo.plot(
    kind="bar",
    figsize=(16,7),
    colormap="Set3"
)

plt.title("Promedio de Estadísticas por Tipo Principal", fontsize=16)
plt.xlabel("Tipo Principal")
plt.ylabel("Promedio")

plt.xticks(rotation=45)

plt.show()

# =========================================================
# 10. POKÉMON LEGENDARIOS VS NO LEGENDARIOS
# =========================================================

plt.figure(figsize=(8,6))

sns.boxplot(
    data=df,
    x="Legendary",
    y="Total",
    palette=["#3498db", "#e74c3c"]
)

plt.title("Total de Estadísticas: Legendarios vs No Legendarios", fontsize=16)
plt.xlabel("Legendario")
plt.ylabel("Total")

plt.show()

# =========================================================
# 11. DISPERSIÓN: VELOCIDAD VS ATAQUE
# =========================================================

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="Speed",
    y="Attack",
    hue="Type 1",
    palette="tab20",
    s=80
)

plt.title("Velocidad vs Ataque", fontsize=16)
plt.xlabel("Velocidad")
plt.ylabel("Ataque")

plt.show()

# =========================================================
# 12. DISPERSIÓN: HP VS DEFENSA
# =========================================================

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="HP",
    y="Defense",
    hue="Legendary",
    palette="Set1",
    s=90
)

plt.title("HP vs Defensa", fontsize=16)
plt.xlabel("HP")
plt.ylabel("Defensa")

plt.show()

# =========================================================
# 13. KDE DISTRIBUCIONAL POR LEGENDARIOS
# =========================================================

plt.figure(figsize=(10,6))

sns.kdeplot(
    data=df,
    x="Total",
    hue="Legendary",
    fill=True,
    palette="coolwarm"
)

plt.title("Distribución KDE del Total según Legendary", fontsize=16)
plt.xlabel("Total")
plt.ylabel("Densidad")

plt.show()

# =========================================================
# 14. VIOLINPLOT DE ATAQUE POR GENERACIÓN
# =========================================================

plt.figure(figsize=(12,6))

sns.violinplot(
    data=df,
    x="Generation",
    y="Attack",
    palette="Spectral"
)

plt.title("Distribución de Ataque por Generación", fontsize=16)
plt.xlabel("Generación")
plt.ylabel("Ataque")

plt.show()

# =========================================================
# 15. PAIRPLOT MULTIVARIADO
# =========================================================

sns.pairplot(
    df[
        [
            "HP",
            "Attack",
            "Defense",
            "Sp. Atk",
            "Sp. Def",
            "Speed",
            "Legendary"
        ]
    ],
    hue="Legendary",
    palette="coolwarm"
)

plt.show()

# =========================================================
# 16. RELACIÓN ENTRE SP. ATK Y SP. DEF
# =========================================================

plt.figure(figsize=(10,6))

sns.regplot(
    data=df,
    x="Sp. Atk",
    y="Sp. Def",
    scatter_kws={"alpha":0.6},
    line_kws={"color":"red"}
)

plt.title("Relación entre Sp. Atk y Sp. Def", fontsize=16)
plt.xlabel("Sp. Atk")
plt.ylabel("Sp. Def")

plt.show()

# =========================================================
# 17. BOXPLOT DE ATAQUE SEGÚN LEGENDARIO
# =========================================================

plt.figure(figsize=(8,6))

sns.boxplot(
    data=df,
    x="Legendary",
    y="Attack",
    palette="pastel"
)

plt.title("Distribución de Ataque según Legendary", fontsize=16)
plt.xlabel("Legendario")
plt.ylabel("Ataque")

plt.show()

# =========================================================
# 18. HISTOGRAMA DE VELOCIDAD
# =========================================================

plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="Speed",
    kde=True,
    bins=25,
    color="orange"
)

plt.title("Distribución de Velocidad", fontsize=16)
plt.xlabel("Velocidad")
plt.ylabel("Frecuencia")

plt.show()

# =========================================================
# 19. DISPERSIÓN TOTAL VS VELOCIDAD
# =========================================================

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="Speed",
    y="Total",
    hue="Generation",
    palette="viridis",
    s=90
)

plt.title("Velocidad vs Total", fontsize=16)
plt.xlabel("Velocidad")
plt.ylabel("Total")

plt.show()

# =========================================================
# 20. JOINTPLOT DE ATAQUE Y DEFENSA
# =========================================================

sns.jointplot(
    data=df,
    x="Attack",
    y="Defense",
    kind="hex",
    cmap="magma"
)

plt.show()
```
